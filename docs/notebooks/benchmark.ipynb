```python
import time
import a5
import h3
import a5_fast

coord = (139.70033506856208, 35.690624903733436)
a5_resolution = 12
h3_resolution = 8
N = 1000
grid_disk_k = 10

polygon = [
    [
        132.3555094, 34.3465028
    ],
    [
        132.4945511, 34.3288744
    ],
    [
        132.5876105, 34.4237528
    ],
    [
        132.6023905, 34.5338573
    ],
    [
        132.4075132, 34.5600083
    ],
    [
        132.2920101, 34.5022848
    ],
    [
        132.2575233, 34.4065923
    ],
    [
        132.3555094, 34.3465028
    ]
]

# a5.lonlat_to_cell
t0 = time.perf_counter()
for _ in range(N):
    cell = a5.lonlat_to_cell(coord, a5_resolution)
t1 = time.perf_counter()
print(f"a5.lonlat_to_cell: {(t1-t0)/N*1e6:.2f} us/call")

# h3.latlng_to_cell
t0 = time.perf_counter()
for _ in range(N):
    cell = h3.latlng_to_cell(coord[1], coord[0], h3_resolution)
t1 = time.perf_counter()
print(f"h3.latlng_to_cell: {(t1-t0)/N*1e6:.2f} us/call")

# a5_fast.lonlat_to_cell
t0 = time.perf_counter()
for _ in range(N):
    cell = a5_fast.lonlat_to_cell(coord[0], coord[1], a5_resolution)
t1 = time.perf_counter()
print(f"a5_fast.lonlat_to_cell: {(t1-t0)/N*1e6:.2f} us/call")

# a5.grid_disk
cell = a5.lonlat_to_cell(coord, a5_resolution)
t0 = time.perf_counter()
for _ in range(N):
    a5.grid_disk(cell, grid_disk_k)
t1 = time.perf_counter()
print(f"a5.grid_disk(k=10): {(t1-t0)/N*1e6:.2f} us/call")

# h3.grid_disk
cell = h3.latlng_to_cell(coord[1], coord[0], h3_resolution)
t0 = time.perf_counter()
for _ in range(N):
    h3.grid_disk(cell, grid_disk_k)
t1 = time.perf_counter()
print(f"h3.grid_disk(k=10): {(t1-t0)/N*1e6:.2f} us/call")

# a5.polygon_to_cells
t0 = time.perf_counter()
for _ in range(N):
    a5.polygon_to_cells(polygon, a5_resolution)
t1 = time.perf_counter()
print(f"a5.polygon_to_cells: {(t1-t0)/N*1e6:.2f} us/call")

# h3.polygon_to_cells
h3_polygon = h3.LatLngPoly(polygon)
t0 = time.perf_counter()
for _ in range(N):
    h3.polygon_to_cells(h3_polygon, h3_resolution)
t1 = time.perf_counter()
print(f"h3.polygon_to_cells: {(t1-t0)/N*1e6:.2f} us/call")
```
